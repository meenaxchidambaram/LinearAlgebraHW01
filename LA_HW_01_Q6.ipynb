# Linear Algebra HW 01 - Part B: Q6

#Imports and Definitions:

```
import numpy as np
```

```
#Set print options for better readability of the output arrays
#this forces NumPy to print floating-point numbers using fixed-point notation (e.g., 0.0001)
#instead of scientific notation (e.g., 1.e-04).
np.set_printoptions(suppress=True, precision=4)
```

Define the matrix A:

```
A = np.array([
    [1, 2, 0, 2, 1],
    [3, 6, 1, 9, 6],
    [2, 4, 1, 7, 5]
])
```

#Compute SVD and Determine Rank



Compute the SVD:

```
# Use an in-built function to compute the SVD
# U: Unitary matrix (m x m). The columns are the left-singular vectors.
# s: 1-D array of singular values of A.
# Vh: Unitary matrix (n x n) (V transpose). The rows are the right-singular vectors.
U, s, Vh = np.linalg.svd(A)
print("U: ",U)
print("s: ",s)
print("Vh: ",Vh)
```

Determine the rank:

```
# 3. Determine the rank (the number of non-zero singular values) from the singular values
tolerance = 1e-10 # We use a small tolerance to account for floating-point inaccuracies.
print("1-D array of singular values of A: ", s)
print("Non-zero values in S: ", s > tolerance) #only count the values that are non-zero
rank = np.sum(s > tolerance) #define rank as number of non singular values
print("Rank of matrix: ",rank)
```

#Compute the bases for the four fundamental subspaces

```
# 4. Compute the bases for the four fundamental subspaces from U and Vh
V = Vh.T # V is the transpose of Vh
m, n = A.shape
print(m,n)
```

Basis for the Column Space C(A) - The first 'rank' columns of U:

```
col_space_basis = U[:, :rank]
```

Basis for the Row Space C(A^T) - The first 'rank' columns of V:

```
row_space_basis = V[:, :rank]
```

Basis for the Null Space N(A) - The last 'n - rank' columns of V:

```
null_space_basis = V[:, rank:]
```

Basis for the Left Null Space N(A^T) - The last 'm - rank' columns of U:

```
left_null_space_basis = U[:, rank:]
```

#Conclusion:

```
print("Original Matrix A:")
print(A)
print("\n" + "-"*40)
print(f"Shape of A: (m, n) = {A.shape}")
print(f"Singular values (s): {s}")
print(f"Rank of A (r): {rank}")
print("-" * 40 + "\n")

print("Basis for the Column Space, C(A)")
print("A basis is formed by the first r columns of U.")
print(f"Dimension = r = {col_space_basis.shape[1]}")
print(col_space_basis)
print("\n" + "-"*40 + "\n")

print("Basis for the Row Space, C(A^T)")
print("A basis is formed by the first r columns of V.")
print(f"Dimension = r = {row_space_basis.shape[1]}")
print(row_space_basis)
print("\n" + "-"*40 + "\n")

print("Basis for the Null Space, N(A)")
print("A basis is formed by the last n-r columns of V.")
print(f"Dimension = n - r = {null_space_basis.shape[1]}")
print(null_space_basis)
print("\n" + "-"*40 + "\n")

print("Basis for the Left Null Space, N(A^T)")
print("A basis is formed by the last m-r columns of U.")
print(f"Dimension = m - r = {left_null_space_basis.shape[1]}")
print(left_null_space_basis)
```
